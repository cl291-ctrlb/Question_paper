```python
# # Using the dataset given
# # ii) conduct a bivariate analysis between marks and study hours visualise  and interpret the relationship
# iii)Perform a multivariate analysis including gender to analyse how gender influences the relationship
# between study hours and marks use seaborns hue parameter to help provide visualisations and insights for each part
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.DataFrame({
    "Marks" : [88, 76, 92, 85, 67, 73, 90, 81, 77, 69],
    "Study_Hours" :  [6, 4, 7, 5.5, 3, 4.5, 6.5, 5, 4, 3.5],
    "Gender" : ['F','M','F','F','M','M','F','M','F','M']
})
sns.lmplot(x="Study_Hours",y="Marks",data=df,height=5,aspect=1.2)  # lmplot ,liner model plot in seaborn,x is dependent var
plt.title("Study_Hours vs  Marks")
plt.show()
```
