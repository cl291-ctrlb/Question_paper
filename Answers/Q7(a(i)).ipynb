```python
#i) Explain how bubble plots can help visualize differences in revenue across categories dynamically.
```

```python
import plotly.express as px
import pandas as pd

# Sample data
data = {
    "Category": ["Electronics", "Smartphones", "Peripherals", "Clothing", "Books"],
    "Revenue": [50000, 75000, 30000, 20000, 15000],
    "Profit": [10000, 20000, 5000, 3000, 2000]
}

df = pd.DataFrame(data)

# Create bubble plot
fig = px.scatter(
    df,
    x="Category",
    y="Revenue",
    size="Profit",          # Bubble size represents Profit
    color="Category",       # Different color for each category
    hover_name="Category",  # Shows category name on hover
    hover_data=["Revenue", "Profit"],  # Show detailed info dynamically
    title="Revenue and Profit Across Categories"
)

fig.show()
```
