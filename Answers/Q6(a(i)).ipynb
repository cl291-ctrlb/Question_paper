```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.DataFrame({
    "Marks" : [88, 76, 92, 85, 67, 73, 90, 81, 77, 69],
    "Study_Hours" :  [6, 4, 7, 5.5, 3, 4.5, 6.5, 5, 4, 3.5],
    "Gender" : ['F','M','F','F','M','M','F','M','F','M']
})
sns.histplot(df["Marks"],kde=True,bins=6,color="skyblue") #kde = kernel density estimate, if true then it gives smooth probability density curve on top of histogram
plt.title("distribution of marks") # bins=6 , number of bins or interval in the histogram, divide dataset into 6 equal width intervals and frequecy data points
plt.xlabel("Marks")
plt.ylabel("frequency")
plt.show()
```
