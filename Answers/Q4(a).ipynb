```python
#matlplotlib
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [4,2,5,4,7]
plt.plot(x,y,marker='o',color = "blue") # 'o' means a circle marker 🔵 will be drawn at every (x, y) point.
                                        # s for square , ^ for triangle, * for star


plt.title ("matplotlib: simple line plot")
plt.xlabel("x axix")
plt.ylabel("y axix")
plt.show()
```

```python
import seaborn as sns
import pandas as pd 
df=pd.DataFrame({
    "Category":["A","B","C","D","E"],
    "Values":[10,20,25,48,78]
})
sns.barplot(x="Category", y="Values",data=df) #a!=A remember this
plt.title("Seaborn : bar plot")
plt.show()
```

```python
#plotly # interesting buddy
import plotly.express as px

df=pd.DataFrame({
    "Fruit":["Apal","Bnana","Orainge","Greyps","H20_melon"],
    "Quantity":[10,20,25,48,78]
})

fig = px.bar(df,x="Fruit",y="Quantity",title="Plotly:Interactive bar chart",color="Fruit")
#color="Fruit"  Assigns different colors to each bar based on the "Fruit" category.

fig.show()
```

```python
# Library	    Nature	                      Best Use Case	                          Example Plot

# Matplotlib	Static, low-level	         Highly customized scientific charts	Line plot
# Seaborn	    Statistical, high-level	     Attractive plots with less code (EDA)	Bar plot, heatmap
# Plotly	    Interactive	Dashboards,      business reports, presentations	Interactive charts
```
