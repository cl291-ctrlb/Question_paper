```python
# # A manufacturing company is analyzing the temperature distribution across a metal sheet during 
# # a healting process . temperature is recorded  at specific points along the x and y axes
# Describe the steps to create a sunburst chart using temperature data where each layer represents different temperature ranges
# categorized by x and y coordinates explain how a sunburst chart can provide insights about the relationships in data
```

```python
import pandas as pd
import plotly.express as px

# Example data
df = pd.DataFrame({
    "X":[0,0,0,5,5,5,10,10,10],
    "Y":[0,5,10,0,5,10,0,5,10],
    "Temp":[25,30,35,28,33,38,41,36,42],
    "Temp Range":["20-30","20-30","30-40","20-30","30-40","30-40","40-50","30-40","40-50"]
})

# Sunburst chart
fig = px.sunburst #function from plotyl express creates a sunburst chart

(df, path=["Temp Range","X","Y"], # defines hierarchy or layers of chart 
        
    values="Temp", #define what size each section of chart represents
                    # Example: Higher Temp = bigger slice

                  color="Temp", # assign color coding based on the temp column 
                                # Each slice gets its color depending on its temperature value

                color_continuous_scale="RdBu", # red to blue (red means higher temp and blue means lower temp)
                  title="Temperature Distribution on Metal Sheet")
fig.show()
```
