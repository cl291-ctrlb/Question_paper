```python
import matplotlib.pyplot as plt
from pywaffle import Waffle
import seaborn as sns
```

```python
data={'ELectronics':30,
      'clothing':20,
        'Groceries':50}
colors=sns.color_palette("Set2",len(data))
fig=plt.figure(
    FigureClass=Waffle,
    rows=10,                 # makes  10x10 grid
    values=data,              # size of each category determines how many squares it gets
    colors=colors,       # assign distinct colors from seaborn palette
    title={'label':'product category distribution','loc':'center'},                      # loc means location of title
    labels=[f"{k} ({v}%)"                    # build text label combining key and value (If k = "A", v = 40 → "A (40%)")
            for k , v in data.items()],                 # makes list of labels
    
    legend={'loc':'upper left','bbox_to_anchor':(1,1)}                   # bbox_to_anchor = "bounding box to anchor" → tells matplotlib where to place the legend box relative to the axes.
                                                                        #It takes a tuple like (x, y) or (x, y, width, height) in axes coordinates.

                                                                        #So the legend will appear at the top-right corner, outside the main chart area.


)
plt.show()
```
