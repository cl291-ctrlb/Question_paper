```python
# Generate a regression plot to visualize the relation between total_bill and tip ,using the sex columns as hue parameter 
#for tips dataset interpret how the regression lines differ for male and female customers
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips") # load_dataset is built in function that lets you quickly load sample dataset for practice
sns.lmplot(x="total_bill",y="tip", hue="sex",data=tips,aspect = 1.5) # lmplot means calls seaborn s linear model plot funcion 

# hue="sex" mean splits data into group based on sex column and helps to compare trends 
#asp

plt.title("Regression plot of total bill vs tip by gender")
plt.show()
```
