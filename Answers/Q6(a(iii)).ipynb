```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.DataFrame({
    "Marks" : [88, 76, 92, 85, 67, 73, 90, 81, 77, 69],
    "Study_Hours" :  [6, 4, 7, 5.5, 3, 4.5, 6.5, 5, 4, 3.5],
    "Gender" : ['F','M','F','F','M','M','F','M','F','M']
})
sns.lmplot(x="Study_Hours",y="Marks",hue="Gender",data=df,height=5,aspect=1.2,markers=["o","s"],palette="Set1") # hue = gender means split data into subgroups based on the gender column
plt.title("Study_Hours vs  Marks by gender") # markers = o for circle,s for square
plt.show()
```
