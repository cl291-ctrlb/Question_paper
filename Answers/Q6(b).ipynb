```python
import folium

# Dataset
city = ['Mumbai', 'Chennai', 'Kolkata', 'Bengaluru']
lat = [19.0760, 13.0827, 22.5726, 12.9716]
lon = [72.8777, 80.2707, 88.3639, 77.5946]
aqi = [160, 120, 180, 140]

# Create map centered on India
m = folium.Map(location=[20.5937, 78.9629], zoom_start=5)

# Add circle markers
for c, lt, ln, q in zip(city, lat, lon, aqi):
    folium.CircleMarker(
        location=[lt, ln],
        radius=q/20,  # radius proportional to AQI
        color="crimson",
        fill=True,
        fill_color="crimson",
        fill_opacity=0.6,
        popup=f"{c} - AQI: {q}"
    ).add_to(m)

# Add legend
legend_html = """
<div style="
position: fixed;
bottom: 50px; left: 50px; width: 150px; height: 120px;
background-color: white; border:2px solid grey; z-index:9999;
font-size:14px; padding:10px;">
<b>Air Quality Index</b><br>
< 100 : Good <br>
100–150 : Moderate <br>
150–200 : Unhealthy <br>
</div>
"""

m.get_root().html.add_child(folium.Element(legend_html))

# Save map
m.save("aqi_map.html")
m
```
